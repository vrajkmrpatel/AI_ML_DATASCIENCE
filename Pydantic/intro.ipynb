```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):
    name:str
    age:int
    city:str

person=Person(name="Vraj", age=23, city="Gandhinagar")
print(person)
```

```python
## Optional field
from typing import Optional

class Employee(BaseModel):
    id:int
    name: str
    department: str
    salary: Optional[float] = None
    is_active: Optional[bool] = True
```

```python
emp1=Employee(id=101, name="Suresh", department="IT")
print(emp1)
```

```python
emp2=Employee(id=2, name="John", department="HR", salary=45000,is_active=False)
print(emp2)
```
